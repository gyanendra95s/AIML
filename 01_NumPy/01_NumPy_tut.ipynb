```python
import numpy as np
import time
```

## NumPy Array are faster than Python List

```python
#executation Performance
size=10_000_000

#Python List
py_list=list(range(size))
start=time.time()
sq_list=[x**2 for x in py_list]
end=time.time()
print(f"python list time= {end-start} seconds")

#NumPy array
np_array=np.array(py_list)
start=time.time()
#vectorization
sq_array=np_array**2
end=time.time()
print(f"numpy array time= {end-start} seconds")
```
